```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

df = pd.read_csv('C:/repos/ICD/IntroCD/dataset/clean/stats_by_player.csv')
win_minutes_relation = pd.read_csv('C:/repos/ICD/IntroCD/dataset/clean/win_minutes_relation.csv')

df = df.merge(win_minutes_relation.drop(columns = 'minutes'), on = 'personName')

df.columns
```

```python
df_c = df[df['position'].str.count('C') > 0]
df_g = df[df['position'].str.count('G') > 0]
df_f = df[df['position'].str.count('F') > 0]

c_features = ['freeThrowsMade', 'reboundsDefensive', 'reboundsOffensive', 'blocks']

g_features = ['threePointersAttempted', 'assists', 'steals', 'turnovers']

f_features = ['threePointersAttempted', 'reboundsDefensive', 'steals', 'turnovers']

df_c = df_c[c_features + ['W']]

df_g = df_g[g_features + ['W']]

df_f = df_f[f_features + ['W']]
```

```python
import statsmodels.api as sm

exog = sm.add_constant(df_c[c_features])
poission_model = sm.GLM(df_c['W'], exog, family=sm.families.Poisson())
result = poission_model.fit()

result.summary()
```
