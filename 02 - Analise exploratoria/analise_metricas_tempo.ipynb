```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

file_path = '../dataset/clean/stats_by_player.csv'

data = pd.read_csv(file_path)
fig, axs = plt.subplots(1, 3, figsize=(18, 6))

sns.scatterplot(x='minutes', y='points', data=data, ax=axs[0])
axs[0].set_title('Distribuição entre Tempo em Quadra e Pontos')
axs[0].set_xlabel('Tempo em Quadra (Minutos)')
axs[0].set_ylabel('Pontos')

sns.scatterplot(x='minutes', y='assists', data=data, ax=axs[1])
axs[1].set_title('Distribuição entre Tempo em Quadra e Assistências')
axs[1].set_xlabel('Tempo em Quadra (Minutos)')
axs[1].set_ylabel('Assistências')

sns.scatterplot(x='minutes', y='reboundsTotal', data=data, ax=axs[2])
axs[2].set_title('Distribuição entre Tempo em Quadra e Rebotes')
axs[2].set_xlabel('Tempo em Quadra (Minutos)')
axs[2].set_ylabel('Rebotes')

plt.tight_layout()
plt.show()

plt.figure(figsize=(10, 6))
sns.histplot(data['minutes'], bins=20, kde=True)
plt.title('Distribuição do Tempo de Jogo de Cada Jogador')
plt.xlabel('Tempo de Jogo (Minutos)')
plt.ylabel('Frequência')
plt.show()
```
