```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

win_minutes_relation = pd.read_csv("../dataset/clean/win_minutes_relation.csv")
win_minutes_relation.columns
```

```python
ax = sns.boxplot(x='W', y='minutes', data=win_minutes_relation)
ax.set(xlabel='Número de vitórias', ylabel='Tempo em quadra (Minutos)')
plt.show()
```

```python
ax = win_minutes_relation.groupby('W')['minutes'].mean().plot(kind='bar')
ax.set(xlabel='Número de vitórias', ylabel='Tempo em quadra (Minutos)')
plt.show()
```

```python
ax = sns.scatterplot(x='W', y='minutes', data=win_minutes_relation)
ax.set(xlabel='Número de vitórias', ylabel='Tempo em quadra (Minutos)')
plt.show()
```
