```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df1 = pd.read_csv('../dataset/raw/regular_season_box_scores_2010_2024_part_1.csv')
df2 = pd.read_csv('../dataset/raw/regular_season_box_scores_2010_2024_part_2.csv')
df3 = pd.read_csv('../dataset/raw/regular_season_box_scores_2010_2024_part_3.csv')
df4 = pd.read_csv('../dataset/raw/regular_season_totals_2010_2024.csv')

df = pd.concat([df1, df2, df3], axis = 0)

df = df.drop(columns = ['jerseyNum', 'comment'], axis = 1)
df['position'] = df.groupby('personName')['position'].transform(lambda x: x.ffill().bfill())
df = df.dropna()
df['minutes'] = df['minutes'].astype(str)
df['minutes'] = df['minutes'].apply(lambda x: int(x.split(':')[0]) + (int(x.split(':')[1])/60))
selection = df4[['GAME_ID', 'WL', 'TEAM_ABBREVIATION']].copy()
selection = selection.rename({'GAME_ID' : 'gameId', 'TEAM_ABBREVIATION' : 'teamTricode'}, axis = 1)
df = pd.merge(df, selection, on = ['gameId', 'teamTricode'])
df
```

```python
player_stats = df.drop(columns = ['game_date', 'gameId', 'matchup', 'teamId', 'teamCity', 'teamTricode', 'teamSlug', 'personId'], axis = 1).copy()
player_stats = player_stats[['personName', 'teamName', 'season_year', 'position', 'minutes']]
player_stats = player_stats.groupby(by = ['personName', 'teamName', 'season_year', 'position']).mean(numeric_only=True).reset_index()
player_stats = player_stats.rename(columns={'minutes': 'mean_minutes_played'})

player_stats_2 = df.drop(columns = ['game_date', 'gameId', 'matchup', 'teamId', 'teamCity', 'teamTricode', 'teamSlug', 'personId'], axis = 1).copy()
player_stats_2 = player_stats_2[['personName', 'teamName', 'season_year', 'position', 'minutes']]
player_stats_2 = player_stats_2.groupby(by = ['personName', 'teamName', 'season_year', 'position']).sum(numeric_only=True).reset_index()
player_stats_2 = player_stats_2.rename(columns={'minutes': 'total_minutes_played'})

player_stats = pd.merge(player_stats, player_stats_2, on=['personName', 'teamName', 'season_year', 'position'], how='inner')
player_stats

player_stats.to_csv('../dataset/clean/player_time_stats.csv', index = False)
```

```python
team_wins = df.drop(columns = ['game_date', 'matchup', 'teamId', 'teamCity', 'teamTricode', 'teamSlug', 'personId'], axis = 1).copy()
team_wins = team_wins[['gameId','teamName', 'season_year', 'WL']]
team_wins = team_wins[team_wins['WL']=='W']
team_wins = team_wins.drop_duplicates(subset=['gameId','teamName', 'season_year', 'WL'])

team_wins = team_wins.groupby(by = ['teamName', 'season_year', 'WL']).size().reset_index(name='wins')[['teamName', 'season_year', 'wins']]
team_wins

#win_minutes_relation = df_dummies.drop(df_dummies[player_stats.WL == 'L'].index).groupby(by = ['personName', 'teamName', 'season_year', 'position']).sum()
#win_minutes_relation
#win_minutes_relation.to_csv('../dataset/clean/win_minutes_relation.csv', index = False)
#player_stats.to_csv('../dataset/clean/stats_by_player.csv', index = False)
```

```python
team_time_played = df.drop(columns = ['game_date', 'matchup', 'teamId', 'teamCity', 'teamTricode', 'teamSlug', 'personId'], axis = 1).copy()
team_time_played = team_time_played[['teamName', 'season_year', 'minutes']]

team_time_played = team_time_played.groupby(by = ['teamName', 'season_year']).sum(numeric_only=True)
team_time_played
```

```python
team_stats = pd.merge(team_time_played, team_wins, on=['teamName', 'season_year'], how='inner')
team_stats

team_stats.to_csv('../dataset/clean/team_time_stats.csv', index = False)
```
